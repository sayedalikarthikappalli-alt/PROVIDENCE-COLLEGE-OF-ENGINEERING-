```python
#load the data.....
import pandas as pd
file = pd.read_excel('/content/Adult_Income_Dataset.xlsx')
```

```python
print(file)
```

```python
# logistic regression works well for numerical data.
file = file.dropna()
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
for column in file.columns:
  if file[column].dtype == object:
    file[column] = le.fit_transform(file[column])
```

```python
x = file.drop('income',axis=1)
y = file['income']
```

```python
#load the packages......
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score,f1_score,recall_score,precision_score,confusion_matrix
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
model1 = LogisticRegression()
model1.fit(x_train,y_train)
y_pred1 = model1.predict(x_test)
accuracy1 = accuracy_score(y_test,y_pred1)
print("******Model 1 (Logistic Regression) Classifiers******")
print("Model 1 Accuracy:",accuracy1)
precision1 = precision_score(y_test,y_pred1)
print("Model 1 Precision:",precision1)
recall1 = recall_score(y_test,y_pred1)
print("Model 1 Recall:",recall1)
f1score1 = f1_score(y_test,y_pred1)
print("Model 1 F1 Score:",f1score1)
print("******Model 2 (Decision Tree) Classifiers******")
model2 = DecisionTreeClassifier()
model2.fit(x_train,y_train)
y_pred2 = model2.predict(x_test)
accuracy2 = accuracy_score(y_test,y_pred2)
print("Model 2 Accuracy:",accuracy2)
precision2 = precision_score(y_test,y_pred2)
print("Precision of Model 2:",precision2)
recall2 = recall_score(y_test,y_pred2)
print("Recall score of Model 2 :",recall2)
f1score2 = f1_score(y_test,y_pred2)
print("F1 score of Model 2 :",f1score2)
```
